# Colab setup, data ingest, splits, normalization, and a tiny CNN

```python
# ==== 0) Setup (Colab) ====
!pip -q install torch torchvision --index-url https://download.pytorch.org/whl/cu121

import os, zipfile, random, math
from pathlib import Path
import torch
from torch.utils.data import DataLoader, random_split
from torchvision import datasets, transforms
import torchvision.utils as vutils
import matplotlib.pyplot as plt

# ==== 1) Unzip your data.zip into /content/data ====
ZIP_PATH = "/content/data.zip"
DATA_DIR = "/content/data"

if os.path.exists(ZIP_PATH):
    os.makedirs(DATA_DIR, exist_ok=True)
    with zipfile.ZipFile(ZIP_PATH, 'r') as zf:
        zf.extractall(DATA_DIR)
else:
    print("⚠️ data.zip not found at /content/. Upload it via the left sidebar > Files.")

# If your archive extracted an extra wrapper folder, try to auto-detect the leaf folder with class subdirs:
def find_leaf_dataset_dir(base):
    base = Path(base)
    # If immediate children look like class folders (contain images), use base
    candidates = [p for p in base.iterdir() if p.is_dir()]
    if len(candidates) == 0:
        return str(base)

    # Heuristic: pick the first directory level that has >=2 subfolders (classes)
    stack = [base]
    while stack:
        d = stack.pop()
        subs = [p for p in d.iterdir() if p.is_dir()]
        if len(subs) >= 2:
            # assume this is the class-level directory if subs don't themselves contain only 1 dir each
            # otherwise keep descending
            # (works well for dumps like data/data/...)
            return str(d)
        stack.extend(subs)
    return str(base)

DATA_DIR = find_leaf_dataset_dir(DATA_DIR)
print("📁 Using dataset dir:", DATA_DIR)
print("📂 Example subfolders:", [p.name for p in Path(DATA_DIR).iterdir() if p.is_dir()][:5])

# ==== 2) Train/Val split folders (optional but handy) ====
# If your data isn't already split into train/val, we'll create an ImageFolder on the whole
# and then split indexes. No need to move files.

# Basic transform for size only (no normalize yet; we’ll compute mean/std first)
IMG_SIZE = 224
base_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
])

whole_ds = datasets.ImageFolder(root=DATA_DIR, transform=base_transform)
num_items = len(whole_ds)
val_ratio = 0.2
val_count = int(math.floor(num_items * val_ratio))
train_count = num_items - val_count
train_ds_base, val_ds_base = random_split(whole_ds, [train_count, val_count])

print(f"🧮 Total: {num_items} | Train: {len(train_ds_base)} | Val: {len(val_ds_base)}")
print("📝 Classes:", whole_ds.classes)

# ==== 3) Compute mean/std on training set (for normalization) ====
def compute_mean_std(dataloader, max_batches=None):
    # Accumulate over batches
    mean = 0.0
    sq_mean = 0.0
    count = 0
    for i, (x, _) in enumerate(dataloader):
        # x: [B, C, H, W], values in [0,1]
        b = x.size(0)
        x_ = x.view(b, x.size(1), -1)  # [B, C, H*W]
        mean += x_.mean(dim=(0,2))
        sq_mean += (x_**2).mean(dim=(0,2))
        count += 1
        if max_batches is not None and (i+1) >= max_batches:
            break
    mean /= count
    sq_mean /= count
    std = torch.sqrt(sq_mean - mean**2)
    return mean, std

tmp_loader = DataLoader(train_ds_base, batch_size=32, shuffle=True, num_workers=2, pin_memory=True)
mean, std = compute_mean_std(tmp_loader, max_batches=50)  # limit for speed if huge
print(f"📏 Mean: {mean.tolist()}")
print(f"📏 Std : {std.tolist()}")

# ==== 4) Final transforms ====
# Augmentations only on TRAIN (flip, rotation, brightness). Validation = resize + normalize only.
train_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.RandomHorizontalFlip(p=0.5),
    transforms.RandomRotation(degrees=15),         # small rotations
    transforms.ColorJitter(brightness=0.2),        # +/- 20% brightness
    transforms.ToTensor(),
    transforms.Normalize(mean=mean.tolist(), std=std.tolist()),
])

val_transform = transforms.Compose([
    transforms.Resize((IMG_SIZE, IMG_SIZE)),
    transforms.ToTensor(),
    transforms.Normalize(mean=mean.tolist(), std=std.tolist()),
])

# Wrap the splits with the new transforms
# (random_split returns Subset; we override its dataset.transform via a tiny adapter)
class SubsetWithTransform(torch.utils.data.Dataset):
    def __init__(self, subset, transform):
        self.subset = subset
        self.transform = transform
        self.classes = subset.dataset.classes
        self.class_to_idx = subset.dataset.class_to_idx
    def __len__(self):
        return len(self.subset)
    def __getitem__(self, idx):
        x, y = self.subset[idx]
        # x currently has base_transform applied; we need the PIL image before transforms
        # To access original path:
        # subset.indices maps to original dataset samples
        original_idx = self.subset.indices[idx]
        path, y2 = self.subset.dataset.samples[original_idx]
        from PIL import Image
        img = Image.open(path).convert("RGB")
        return self.transform(img), y2

train_ds = SubsetWithTransform(train_ds_base, train_transform)
val_ds   = SubsetWithTransform(val_ds_base,   val_transform)

# ==== 5) DataLoaders ====
BATCH_SIZE = 32
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=2, pin_memory=True)

# ==== 6) Quick visual sanity check (unnormalized preview) ====
# We'll preview a few items by undoing normalization
inv_mean = torch.tensor(mean).view(3,1,1)
inv_std  = torch.tensor(std).view(3,1,1)

def denorm(img_t):
    return (img_t * inv_std + inv_mean).clamp(0,1)

images, labels = next(iter(train_loader))
grid = vutils.make_grid(denorm(images[:16]), nrow=8, padding=2)
plt.figure(figsize=(12, 4))
plt.axis('off')
plt.imshow(grid.permute(1,2,0).cpu().numpy())
plt.title("Sample batch (after augmentations, de-normalized for display)")
plt.show()

# ==== 7) You’re ready to train ====
# Example model stub:
import torch.nn as nn
import torch.nn.functional as F

class TinyCNN(nn.Module):
    def __init__(self, num_classes):
        super().__init__()
        self.conv1 = nn.Conv2d(3, 16, 3, padding=1)
        self.conv2 = nn.Conv2d(16, 32, 3, padding=1)
        self.pool  = nn.MaxPool2d(2,2)
        self.head  = nn.Linear(32*(IMG_SIZE//4)*(IMG_SIZE//4), num_classes)
    def forward(self, x):
        x = self.pool(F.relu(self.conv1(x)))
        x = self.pool(F.relu(self.conv2(x)))
        x = x.view(x.size(0), -1)
        return self.head(x)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = TinyCNN(num_classes=len(whole_ds.classes)).to(device)
print("✅ Preprocessing ready. Model stub created. Start training with your loop/optimizer.")
```

# Training loop, transfer models, and evaluation

```python
import os, time, copy
import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim.lr_scheduler import CosineAnnealingLR
from torchvision import models
from torch.cuda.amp import GradScaler, autocast

try:
    from sklearn.metrics import classification_report, confusion_matrix
    import numpy as np
    import matplotlib.pyplot as plt
    SKLEARN_OK = True
except Exception:
    SKLEARN_OK = False
    print("sklearn/matplotlib not available; skipping confusion matrix.")

# ---------- Safety checks ----------
assert 'train_loader' in globals() and 'val_loader' in globals(), "Run the preprocessing cell first to create DataLoaders."
assert 'whole_ds' in globals(), "Need whole_ds for class names."
assert 'IMG_SIZE' in globals(), "Need IMG_SIZE from preprocessing cell."
assert 'device' in globals(), "Need device from preprocessing cell (cuda/cpu)."

NUM_CLASSES = len(whole_ds.classes)
CLASS_NAMES = whole_ds.classes
SAVE_DIR = "/content/checkpoints"
os.makedirs(SAVE_DIR, exist_ok=True)

# === 1) Baseline custom CNN (same idea as earlier TinyCNN but packaged) ===
class BaselineCNN(nn.Module):
    def __init__(self, num_classes, img_size=224):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1), nn.ReLU(inplace=True),
            nn.MaxPool2d(2,2),  # /2
            nn.Conv2d(32, 64, 3, padding=1), nn.ReLU(inplace=True),
            nn.MaxPool2d(2,2),  # /4
            nn.Conv2d(64, 128, 3, padding=1), nn.ReLU(inplace=True),
            nn.MaxPool2d(2,2),  # /8
            nn.Conv2d(128, 128, 3, padding=1), nn.ReLU(inplace=True),
            nn.AdaptiveAvgPool2d((1,1))
        )
        self.head = nn.Sequential(
            nn.Flatten(),
            nn.Dropout(0.25),
            nn.Linear(128, 256), nn.ReLU(inplace=True),
            nn.Dropout(0.25),
            nn.Linear(256, num_classes)
        )

    def forward(self, x):
        x = self.features(x)
        x = self.head(x)
        return x

# === 2) Transfer Learning Model Factory (ResNet18 / VGG16) ===
def build_model(name: str, num_classes: int, pretrained=True, freeze_backbone=True):
    name = name.lower()
    if name in ("resnet18", "resnet"):
        m = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1 if pretrained else None)
        in_feats = m.fc.in_features
        m.fc = nn.Linear(in_feats, num_classes)
        if freeze_backbone:
            for p in list(m.parameters())[:-2]:  # freeze most layers except final fc params
                p.requires_grad = False
        return m

    if name in ("vgg16", "vgg"):
        m = models.vgg16(weights=models.VGG16_Weights.IMAGENET1K_V1 if pretrained else None)
        in_feats = m.classifier[-1].in_features
        # replace classifier tail
        m.classifier[-1] = nn.Linear(in_feats, num_classes)
        if freeze_backbone:
            for p in m.features.parameters():
                p.requires_grad = False
        return m

    if name in ("baseline", "custom", "tiny"):
        return BaselineCNN(num_classes=num_classes, img_size=IMG_SIZE)

    raise ValueError(f"Unknown model name: {name}")

# === 3) Train / Evaluate helpers ===
def epoch_run(model, loader, criterion, optimizer=None, scaler=None):
    is_train = optimizer is not None
    model.train(is_train)

    running_loss = 0.0
    running_correct = 0
    total = 0

    for inputs, targets in loader:
        inputs, targets = inputs.to(device, non_blocking=True), targets.to(device, non_blocking=True)

        with autocast(enabled=True):
            outputs = model(inputs)
            loss = criterion(outputs, targets)

        if is_train:
            optimizer.zero_grad(set_to_none=True)
            if scaler is not None:
                # backward with scaling
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
            else:
                loss.backward()
                optimizer.step()

        _, preds = torch.max(outputs, 1)
        running_loss += loss.item() * inputs.size(0)
        running_correct += (preds == targets).sum().item()
        total += inputs.size(0)

    avg_loss = running_loss / max(total, 1)
    avg_acc = running_correct / max(total, 1)
    return avg_loss, avg_acc

def train_model(model, train_loader, val_loader, epochs=10, lr=1e-3, wd=1e-4, use_scheduler=True):
    model = model.to(device)
    criterion = nn.CrossEntropyLoss()

    # Only optimize parameters that require grad (handles frozen backbones)
    optimizer = optim.AdamW(filter(lambda p: p.requires_grad, model.parameters()), lr=lr, weight_decay=wd)
    scheduler = CosineAnnealingLR(optimizer, T_max=epochs) if use_scheduler else None
    scaler = GradScaler(enabled=torch.cuda.is_available())

    best_wts = copy.deepcopy(model.state_dict())
    best_acc = 0.0
    history = {"train_loss": [], "train_acc": [], "val_loss": [], "val_acc": []}

    print(f"🚀 Start training for {epochs} epochs...")
    for epoch in range(1, epochs+1):
        t0 = time.time()
        tr_loss, tr_acc = epoch_run(model, train_loader, criterion, optimizer, scaler)
        va_loss, va_acc = epoch_run(model, val_loader, criterion, optimizer=None)

        if scheduler: scheduler.step()

        history["train_loss"].append(tr_loss)
        history["train_acc"].append(tr_acc)
        history["val_loss"].append(va_loss)
        history["val_acc"].append(va_acc)

        if va_acc > best_acc:
            best_acc = va_acc
            best_wts = copy.deepcopy(model.state_dict())
            torch.save(best_wts, os.path.join(SAVE_DIR, "best.pt"))

        dt = time.time()-t0
        print(f"Epoch {epoch:02d}/{epochs} | "
              f"train: loss {tr_loss:.4f} acc {tr_acc:.4f} | "
              f"val: loss {va_loss:.4f} acc {va_acc:.4f} | "
              f"{dt:.1f}s")

    print(f"✅ Best val acc: {best_acc:.4f}. Best weights saved to {os.path.join(SAVE_DIR,'best.pt')}")
    model.load_state_dict(best_wts)
    return model, history

def evaluate(model, loader):
    model.eval()
    all_preds, all_tgts = [], []
    with torch.no_grad():
        for x, y in loader:
            x = x.to(device)
            logits = model(x)
            preds = logits.argmax(1).cpu().tolist()
            all_preds += preds
            all_tgts += y.tolist()

    if SKLEARN_OK:
        print("\nClassification report:\n")
        print(classification_report(all_tgts, all_preds, target_names=CLASS_NAMES))
        cm = confusion_matrix(all_tgts, all_preds)
        fig, ax = plt.subplots(figsize=(6,6))
        im = ax.imshow(cm, interpolation='nearest')
        ax.set_title("Confusion Matrix")
        ax.set_xticks(np.arange(len(CLASS_NAMES))); ax.set_xticklabels(CLASS_NAMES, rotation=45, ha='right')
        ax.set_yticks(np.arange(len(CLASS_NAMES))); ax.set_yticklabels(CLASS_NAMES)
        for i in range(cm.shape[0]):
            for j in range(cm.shape[1]):
                ax.text(j, i, cm[i, j], ha="center", va="center")
        fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
        plt.tight_layout()
        plt.show()
    else:
        correct = sum(int(p==t) for p,t in zip(all_preds, all_tgts))
        print(f"Accuracy: {correct/len(all_tgts):.4f} (skipping detailed report; install scikit-learn to see it)")

# === 4) Choose and train a model ===
# Options: "baseline" (custom CNN), "resnet18" (transfer), "vgg16" (transfer)
MODEL_NAME = "baseline"     # <- change to 'resnet18' or 'vgg16' to compare
PRETRAINED = True           # only used for resnet/vgg
FREEZE_BACKBONE = True      # set False to fine-tune entire network

model = build_model(MODEL_NAME, NUM_CLASSES, pretrained=PRETRAINED, freeze_backbone=FREEZE_BACKBONE)
model, hist = train_model(model, train_loader, val_loader, epochs=12, lr=3e-4, wd=1e-4, use_scheduler=True)

# Evaluate and visualize
evaluate(model, val_loader)

# === 5) (Optional) Quick training curves ===
try:
    plt.figure(figsize=(6,4)); plt.plot(hist["train_loss"], label="train_loss"); plt.plot(hist["val_loss"], label="val_loss"); plt.legend(); plt.title("Loss"); plt.show()
    plt.figure(figsize=(6,4)); plt.plot(hist["train_acc"], label="train_acc"); plt.plot(hist["val_acc"], label="val_acc"); plt.legend(); plt.title("Accuracy"); plt.show()
except Exception as e:
    pass
```

# Scaffold a YOLO dataset from class folders

```python
import os, glob, random, shutil
from pathlib import Path

# ======== INPUT: your classification dataset root ========
CLS_ROOT = Path("/content/data")  # /content/data/<class>/*.jpg
DET_ROOT = Path("/content/waste_det")
VAL_RATIO = 0.2

# Build YOLO dirs
for p in ["train/images","train/labels","val/images","val/labels"]:
    (DET_ROOT/p).mkdir(parents=True, exist_ok=True)

# Collect class folders and images
classes = [d.name for d in CLS_ROOT.iterdir() if d.is_dir()]
classes.sort()
class_to_id = {c:i for i,c in enumerate(classes)}
print("Classes:", classes)

IMG_EXTS = {".jpg",".jpeg",".png",".bmp",".tif",".tiff",".webp"}
all_images = []
for c in classes:
    imgs = [Path(p) for p in glob.glob(str(CLS_ROOT/c/"**/*"), recursive=True)
            if Path(p).is_file() and Path(p).suffix.lower() in IMG_EXTS]
    all_images += [(c, p) for p in imgs]

print(f"Found {len(all_images)} images total from classification folders.")

# Split train/val per class (stratified-ish)
train_pairs, val_pairs = [], []
random.seed(42)
for c in classes:
    c_imgs = [p for cc,p in all_images if cc==c]
    random.shuffle(c_imgs)
    k = max(1, int(len(c_imgs)*VAL_RATIO))
    val_pairs += [(c,p) for p in c_imgs[:k]]
    train_pairs += [(c,p) for p in c_imgs[k:]]

print(f"Split -> Train: {len(train_pairs)}, Val: {len(val_pairs)}")

def copy_with_placeholder_label(split_pairs, split_name):
    """Copy images and create labels.
       For the FIRST image of each class we create a placeholder full-image box so YOLO sees at least one positive.
       Other images get empty labels (negative samples). Replace later with REAL annotations.
    """
    made_pos_for = set()
    for c, img_path in split_pairs:
        cls_id = class_to_id[c]
        # destination image path mirrors any nesting via basename only (simple)
        dst_img = DET_ROOT/f"{split_name}/images"/img_path.name
        dst_lbl = DET_ROOT/f"{split_name}/labels"/(img_path.stem + ".txt")
        shutil.copy2(img_path, dst_img)

        if c not in made_pos_for:
            # Placeholder: one big box per class on one image, normalized coords (cx, cy, w, h)
            # Here we set a slightly smaller than full frame box to avoid edge issues
            with open(dst_lbl, "w") as f:
                f.write(f"{cls_id} 0.5 0.5 0.9 0.9\n")
            made_pos_for.add(c)
        else:
            # Empty label file (negative sample)
            open(dst_lbl, "w").close()

copy_with_placeholder_label(train_pairs, "train")
copy_with_placeholder_label(val_pairs,   "val")

print("✅ Scaffolded YOLO dataset at:", DET_ROOT)
print("NOTE: These labels are placeholders. Replace with REAL bounding boxes for meaningful training.")
```

# Conditional DCGAN architecture and training setup

```python
import torch.nn as nn
import torchvision.utils as vutils
import matplotlib.pyplot as plt

nz, nc, ngf, ndf, embed_dim = 100, 3, 64, 64, 32

def weights_init(m):
    name = m.__class__.__name__
    if "Conv" in name:
        nn.init.normal_(m.weight, 0.0, 0.02)
    elif "BatchNorm" in name:
        nn.init.normal_(m.weight, 1.0, 0.02)
        nn.init.zeros_(m.bias)

class CondGen(nn.Module):
    def __init__(self):
        super().__init__()
        self.embed = nn.Embedding(num_classes, embed_dim)
        self.net = nn.Sequential(
            nn.ConvTranspose2d(nz+embed_dim, ngf*8, 4, 1, 0, bias=False),
            nn.BatchNorm2d(ngf*8), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*8, ngf*4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf*4), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*4, ngf*2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf*2), nn.ReLU(True),
            nn.ConvTranspose2d(ngf*2, ngf,   4, 2, 1, bias=False),
            nn.BatchNorm2d(ngf), nn.ReLU(True),
            nn.ConvTranspose2d(ngf, nc, 3, 1, 1, bias=False),
            nn.Tanh()
        )
    def forward(self, z, y):
        e = self.embed(y)              # (B, embed_dim)
        zc = torch.cat([z, e], 1)      # (B, nz+embed_dim)
        zc = zc.unsqueeze(-1).unsqueeze(-1)
        return self.net(zc)

class CondDisc(nn.Module):
    """Size-agnostic: broadcasts label channel to match x's HxW."""
    def __init__(self):
        super().__init__()
        self.embed = nn.Embedding(num_classes, 1)
        self.net = nn.Sequential(
            nn.Conv2d(nc+1, ndf, 4, 2, 1, bias=False), nn.LeakyReLU(0.2, True),
            nn.Conv2d(ndf, ndf*2, 4, 2, 1, bias=False), nn.BatchNorm2d(ndf*2), nn.LeakyReLU(0.2, True),
            nn.Conv2d(ndf*2, ndf*4, 4, 2, 1, bias=False), nn.BatchNorm2d(ndf*4), nn.LeakyReLU(0.2, True),
            nn.Conv2d(ndf*4, 1, 4, 1, 0, bias=False),
        )
    def forward(self, x, y):
        B, C, H, W = x.shape
        e = self.embed(y).view(B,1,1,1).expand(B,1,H,W)
        out = self.net(torch.cat([x, e], 1))   # (B,1,h,w)
        return out.mean((2,3)).squeeze(1)      # (B,)

G = CondGen().to(device).apply(weights_init)
D = CondDisc().to(device).apply(weights_init)

criterion = nn.BCEWithLogitsLoss()
optG = torch.optim.Adam(G.parameters(), lr=2e-4, betas=(0.5, 0.999))
optD = torch.optim.Adam(D.parameters(), lr=2e-4, betas=(0.5, 0.999))

# fixed preview
fixed_labels = torch.arange(min(num_classes, 8), device=device)
fixed_noise  = torch.randn(len(fixed_labels), nz, device=device)

def show_grid(tg, title=None):
    img = tg.permute(1,2,0).detach().cpu().numpy()
    plt.figure(figsize=(8,4)); plt.axis("off");
    if title: plt.title(title)
    plt.imshow(img); plt.show()
```

# Conditional DCGAN training loop

```python
from tqdm.auto import tqdm

epochs = 12
G.train(); D.train()

for ep in range(1, epochs+1):
    for real, y in tqdm(gan_loader, desc=f"cDCGAN {ep}/{epochs}"):
        real, y = real.to(device), y.to(device)
        b = real.size(0)

        # -- D --
        z = torch.randn(b, nz, device=device)
        fake = G(z, y).detach()
        d_real = D(real, y)
        d_fake = D(fake, y)
        lossD = criterion(d_real, torch.ones_like(d_real)) + \
                criterion(d_fake, torch.zeros_like(d_fake))
        optD.zero_grad(); lossD.backward(); optD.step()

        # -- G --
        z = torch.randn(b, nz, device=device)
        fake = G(z, y)
        d_fake = D(fake, y)
        lossG = criterion(d_fake, torch.ones_like(d_fake))
        optG.zero_grad(); lossG.backward(); optG.step()

    with torch.no_grad():
        samples = G(fixed_noise, fixed_labels).cpu()*0.5 + 0.5
        grid = vutils.make_grid(samples, nrow=len(fixed_labels))
    print(f"Epoch {ep}: lossD {lossD.item():.3f}  lossG {lossG.item():.3f}")
    show_grid(grid, "GAN samples (classes 0..7)")
```

# Denoising autoencoder (DAE) with noisy–clean pairs and inline previews

```python
# DAE: dataset -> model -> training (with inline preview)
import io, random, math
from pathlib import Path
import torch, torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torchvision import datasets, transforms
from torchvision.transforms import functional as TF
import matplotlib.pyplot as plt
from PIL import Image, ImageFilter

# -------- config --------
DATA_DIR   = Path("/content/data")   # your class-folder root
IMG_SIZE   = 224
BATCH_SIZE = 32
EPOCHS     = 8
LR         = 1e-3
WEIGHTS    = Path("/content/dae_best.pt")

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("device:", device)

# -------- noisy→clean dataset --------
class NoisyClean(Dataset):
    def __init__(self, root: Path, img_size=224):
        self.files = []
        exts = {".jpg",".jpeg",".png",".bmp",".tif",".tiff",".webp"}
        for c in sorted([d for d in root.iterdir() if d.is_dir()]):
            self.files += [p for p in c.rglob("*") if p.suffix.lower() in exts]
        assert len(self.files), f"No images found under {root}"
        self.resize  = transforms.Resize((img_size, img_size))
        self.to_tensor = transforms.ToTensor()

    def add_noise(self, pil: Image.Image) -> Image.Image:
        # 1) random gaussian blur (simulates defocus/dirty lens)
        if random.random() < 0.7:
            pil = pil.filter(ImageFilter.GaussianBlur(radius=random.choice([0,0.6,1.2])))
        # 2) jpeg compression artifacts
        if random.random() < 0.5:
            buf = io.BytesIO()
            pil.save(buf, format="JPEG", quality=random.randint(40,80))
            buf.seek(0)
            pil = Image.open(buf).convert("RGB")
        # 3) add gaussian noise + mild brightness/contrast jitter (in tensor space)
        t = self.to_tensor(pil)
        t = (t + torch.randn_like(t)*0.06).clamp(0,1)
        if random.random() < 0.5:
            t = TF.adjust_brightness(t, 1.0 + random.uniform(-0.2, 0.2))
        if random.random() < 0.5:
            t = TF.adjust_contrast(t,  1.0 + random.uniform(-0.2, 0.2))
        return transforms.ToPILImage()(t)

    def __len__(self): return len(self.files)

    def __getitem__(self, idx):
        p = self.files[idx]
        clean = Image.open(p).convert("RGB")
        noisy = self.add_noise(clean.copy())
        clean = self.to_tensor(self.resize(clean))
        noisy = self.to_tensor(self.resize(noisy))
        return noisy, clean

dae_ds = NoisyClean(DATA_DIR, IMG_SIZE)
dae_dl = DataLoader(dae_ds, batch_size=BATCH_SIZE, shuffle=True, num_workers=2, pin_memory=True)
print("dae samples:", len(dae_ds))

# -------- small UNet-ish model --------
def conv_block(c_in, c_out):
    return nn.Sequential(
        nn.Conv2d(c_in, c_out, 3, padding=1), nn.ReLU(inplace=True),
        nn.Conv2d(c_out, c_out, 3, padding=1), nn.ReLU(inplace=True),
    )

class SmallUNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.enc1 = conv_block(3, 32)
        self.down1 = nn.Conv2d(32, 32, 3, stride=2, padding=1)     # /2
        self.enc2 = conv_block(32, 64)
        self.down2 = nn.Conv2d(64, 64, 3, stride=2, padding=1)     # /4
        self.bott = conv_block(64, 128)
        self.up2  = nn.ConvTranspose2d(128, 64, 2, stride=2)       # x2
        self.dec2 = conv_block(64+64, 64)
        self.up1  = nn.ConvTranspose2d(64, 32, 2, stride=2)        # x2
        self.dec1 = conv_block(32+32, 32)
        self.out  = nn.Conv2d(32, 3, 1)
        self.act  = nn.Sigmoid()

    def forward(self, x):
        e1 = self.enc1(x)
        e2 = self.enc2(self.down1(e1))
        b  = self.bott(self.down2(e2))
        d2 = self.up2(b)
        d2 = torch.cat([d2, e2], dim=1)
        d2 = self.dec2(d2)
        d1 = self.up1(d2)
        d1 = torch.cat([d1, e1], dim=1)
        d1 = self.dec1(d1)
        y  = self.act(self.out(d1))
        return y

dae = SmallUNet().to(device)
opt = torch.optim.AdamW(dae.parameters(), lr=LR, weight_decay=1e-5)
loss_fn = nn.L1Loss()  # MAE is robust to outliers

def psnr(batch_pred, batch_gt, eps=1e-8):
    # inputs in [0,1]
    mse = torch.mean((batch_pred - batch_gt)**2, dim=(1,2,3)) + eps
    return 10.0 * torch.log10(1.0 / mse)

# -------- train --------
best_psnr = -1.0
for ep in range(1, EPOCHS+1):
    dae.train()
    run_loss, run_psnr, n = 0.0, 0.0, 0
    for noisy, clean in dae_dl:
        noisy, clean = noisy.to(device), clean.to(device)
        pred = dae(noisy)
        loss = loss_fn(pred, clean)
        opt.zero_grad(); loss.backward(); opt.step()
        with torch.no_grad():
            run_loss += loss.item() * noisy.size(0)
            run_psnr += psnr(pred.detach().clamp(0,1), clean).sum().item()
            n += noisy.size(0)
    avg_loss = run_loss / n
    avg_psnr = run_psnr / n
    print(f"Epoch {ep}/{EPOCHS}  L1:{avg_loss:.4f}  PSNR:{avg_psnr:.2f} dB")

    # quick preview (first 4)
    dae.eval()
    noisy, clean = next(iter(dae_dl))
    with torch.no_grad():
        den = dae(noisy.to(device)).cpu().clamp(0,1)
    k = min(4, noisy.size(0))
    fig, axes = plt.subplots(k, 3, figsize=(8, 2.2*k))
    for i in range(k):
        axes[i,0].imshow(noisy[i].permute(1,2,0).clamp(0,1)); axes[i,0].set_title("noisy");   axes[i,0].axis("off")
        axes[i,1].imshow(den[i].permute(1,2,0));              axes[i,1].set_title("denoised");axes[i,1].axis("off")
        axes[i,2].imshow(clean[i].permute(1,2,0));            axes[i,2].set_title("clean");   axes[i,2].axis("off")
    plt.tight_layout(); plt.show()

    # save best by PSNR
    if avg_psnr > best_psnr:
        best_psnr = avg_psnr
        torch.save(dae.state_dict(), WEIGHTS)
        print(f"✅ saved best to {WEIGHTS} (PSNR {best_psnr:.2f} dB)")
```

# Denoising at inference and offline cleaning

```python

import torch.nn as nn

class DAEWrapper(nn.Module):
    def __init__(self, dae, classifier, freeze_dae=True):
        super().__init__()
        self.dae = dae.eval()
        if freeze_dae:
            for p in self.dae.parameters(): p.requires_grad = False
        self.classifier = classifier
    def forward(self, x):
        with torch.no_grad():
            x = self.dae(x).clamp(0,1)
        return self.classifier(x)


dae2 = SmallUNet().to(device)
dae2.load_state_dict(torch.load(WEIGHTS, map_location=device))
dae2.eval()


from pathlib import Path
from torchvision.utils import save_image

def clean_dataset(in_root: Path, out_root: Path, batch_size=32):
    exts = {".jpg",".jpeg",".png",".bmp",".tif",".tiff",".webp"}
    files = []
    for c in sorted([d for d in in_root.iterdir() if d.is_dir()]):
        files += [p for p in c.rglob("*") if p.suffix.lower() in exts]
    assert files, f"No images in {in_root}"

    tf = transforms.Compose([transforms.Resize((IMG_SIZE, IMG_SIZE)), transforms.ToTensor()])
    # simple batched loop
    out_root.mkdir(parents=True, exist_ok=True)
    dae2.eval()
    for c in sorted({p.parent.name for p in files}):
        (out_root/c).mkdir(parents=True, exist_ok=True)

    for i in range(0, len(files), batch_size):
        batch = files[i:i+batch_size]
        imgs = [tf(Image.open(p).convert("RGB")) for p in batch]
        x = torch.stack(imgs).to(device)
        with torch.no_grad():
            den = dae2(x).clamp(0,1).cpu()
        for p, img in zip(batch, den):
            save_image(img, (out_root/p.parent.name/f"{p.stem}_clean.png").as_posix())

    print(f"✅ Cleaned dataset saved to: {out_root}")

```

# Image–text dataset wrapper with smart text fallback

```python
class ImageTextFolder(Dataset):
    def __init__(self, root: Path, transform, fallback_template="a photo of {cls}"):
        self.base = datasets.ImageFolder(str(root), transform=None)
        self.samples = self.base.samples
        self.classes = self.base.classes
        self.class_to_idx = self.base.class_to_idx
        self.transform = transform
        self.fallback_template = fallback_template

    def _derive_text(self, path: str, label: int) -> str:
        p = Path(path)
        rel = p.relative_to(DATA_DIR).as_posix()
        # priority: exact relpath -> basename -> class -> fallback
        if rel in DESC_BY_RELPATH: return DESC_BY_RELPATH[rel]
        if p.name in DESC_BY_BASENAME: return DESC_BY_BASENAME[p.name]
        cls = self.classes[label]
        if cls in DESC_BY_CLASS: return DESC_BY_CLASS[cls]
        return self.fallback_template.format(cls=cls.replace("_", " "))

    def __len__(self): return len(self.samples)

    def __getitem__(self, idx):
        path, y = self.samples[idx]
        img = Image.open(path).convert("RGB")
        x = self.transform(img)
        txt = self._derive_text(path, y)
        return x, txt, y
```

# Freeze encoders and fuse image + text

```python
class MeanPooler(nn.Module):
    """Mask-aware mean pooling for Transformer outputs."""
    def forward(self, last_hidden_state, attention_mask):
        # last_hidden_state: (B,T,H), attention_mask: (B,T)
        mask = attention_mask.unsqueeze(-1).float()           # (B,T,1)
        summed = (last_hidden_state * mask).sum(dim=1)        # (B,H)
        counted = mask.sum(dim=1).clamp(min=1e-6)             # (B,1)
        return summed / counted

# --- encoders ---
vision = models.resnet18(weights=models.ResNet18_Weights.IMAGENET1K_V1)
vision.fc = nn.Identity()
VISION_DIM = 512

text_encoder = AutoModel.from_pretrained(TEXT_MODEL_NAME)
TEXT_DIM = text_encoder.config.hidden_size
pooler = MeanPooler()

# freeze encoders for the first run
for p in vision.parameters(): p.requires_grad = False
for p in text_encoder.parameters(): p.requires_grad = False

# --- fusion head ---
class ImageTextClassifier(nn.Module):
    def __init__(self, vision, text_encoder, pooler, vision_dim, text_dim, num_classes):
        super().__init__()
        self.vision = vision
        self.text_encoder = text_encoder
        self.pooler = pooler
        self.head = nn.Sequential(
            nn.Linear(vision_dim + text_dim, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.25),
            nn.Linear(512, num_classes)
        )
    def forward(self, images, input_ids, attention_mask):
        v = self.vision(images)  # (B, VISION_DIM)
        t_out = self.text_encoder(input_ids=input_ids, attention_mask=attention_mask)
        t = self.pooler(t_out.last_hidden_state, attention_mask)  # (B, TEXT_DIM)
        z = torch.cat([v, t], dim=1)
        return self.head(z)

model = ImageTextClassifier(vision, text_encoder, pooler, VISION_DIM, TEXT_DIM, NUM_CLASSES).to(device)
```

# Train the multimodal head with AMP and save the best

```python
import time
from torch.cuda.amp import GradScaler, autocast

criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.AdamW(model.head.parameters(), lr=3e-4, weight_decay=1e-4)
scaler = GradScaler(enabled=torch.cuda.is_available())

def run_epoch(loader, train=True):
    model.train(train)
    total_loss, correct, total = 0.0, 0, 0
    for images, input_ids, attention_mask, y in loader:
        images = images.to(device, non_blocking=True)
        input_ids = input_ids.to(device, non_blocking=True)
        attention_mask = attention_mask.to(device, non_blocking=True)
        y = y.to(device, non_blocking=True)

        with autocast(enabled=torch.cuda.is_available()):
            logits = model(images, input_ids, attention_mask)
            loss = criterion(logits, y)

        if train:
            optimizer.zero_grad(set_to_none=True)
            scaler.scale(loss).backward()
            scaler.step(optimizer)
            scaler.update()

        total_loss += loss.item() * images.size(0)
        pred = logits.argmax(1)
        correct += (pred == y).sum().item()
        total += images.size(0)

    return total_loss / max(total,1), correct / max(total,1)

EPOCHS = 6
best_acc = 0.0
for ep in range(1, EPOCHS+1):
    t0 = time.time()
    tr_loss, tr_acc = run_epoch(train_loader, train=True)
    va_loss, va_acc = run_epoch(val_loader, train=False)
    dt = time.time() - t0
    print(f"Epoch {ep}/{EPOCHS}  "
          f"train: loss {tr_loss:.4f} acc {tr_acc:.4f} |  "
          f"val: loss {va_loss:.4f} acc {va_acc:.4f}  ({dt:.1f}s)")
    if va_acc > best_acc:
        best_acc = va_acc
        torch.save(model.state_dict(), "/content/mm_best.pt")
        print("  ✅ saved /content/mm_best.pt")
print("Best val acc:", best_acc)
```

# Fine-tune encoders (last ResNet block + full DistilBERT)

```python
for name, p in model.vision.named_parameters():
    if name.startswith("layer4"):
        p.requires_grad = True
for p in model.text_encoder.parameters():
    p.requires_grad = True

optimizer = torch.optim.AdamW(filter(lambda p: p.requires_grad, model.parameters()),
                              lr=1e-5, weight_decay=1e-4)

for ep in range(1, 3+1):
    tr_loss, tr_acc = run_epoch(train_loader, train=True)
    va_loss, va_acc = run_epoch(val_loader, train=False)
    print(f"[FT] Epoch {ep}  train acc {tr_acc:.4f} | val acc {va_acc:.4f}")
```

# Flask+Waitress CLIP app with TryCloudflare

```python
# 🔁 Reliable relaunch: Flask + Waitress + zero-shot CLIP + TryCloudflare (keeps tunnel alive)
import sys, subprocess, os, re, socket, threading, time, stat, platform, urllib.request
from pathlib import Path
from typing import List, Tuple
from PIL import Image
from flask import Flask, request, jsonify, render_template_string

# --- ensure deps ---
try:
    from waitress import serve
except Exception:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "waitress==3.0.0"])
    from waitress import serve
try:
    import torch
    from transformers import CLIPModel, CLIPProcessor
except Exception:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "torch>=2.2.0", "transformers==4.43.3", "pillow>=10.3.0"])
    import torch
    from transformers import CLIPModel, CLIPProcessor

# ------------------ Config ------------------
WASTE_TYPES = ["cardboard","glass","metal","paper","plastic","trash"]
PROMPTS = [
    "a photo of cardboard or a cardboard box",
    "a photo of a glass bottle or glass container",
    "a photo of a metal can or metal container",
    "a photo of paper or paper packaging",
    "a photo of a plastic bottle or plastic packaging",
    "a photo of mixed waste or trash",
]
TOPK = len(WASTE_TYPES)
START_PORT = 7860
TUNNEL_TIMEOUT = 45  # a bit longer to be safe

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# ------------------ Model ------------------
print("Loading CLIP (openai/clip-vit-base-patch32)…")
clip_model = CLIPModel.from_pretrained("openai/clip-vit-base-patch32").to(device).eval()
clip_proc  = CLIPProcessor.from_pretrained("openai/clip-vit-base-patch32")
print("✔ CLIP loaded.")

@torch.inference_mode()
def predict_pil(pil_img: Image.Image):
    inputs = clip_proc(text=PROMPTS, images=pil_img, return_tensors="pt", padding=True).to(device)
    out = clip_model(**inputs)
    probs = out.logits_per_image.softmax(dim=-1).squeeze(0)
    vals, idxs = torch.topk(probs, k=min(TOPK, len(WASTE_TYPES)))
    topk = [(WASTE_TYPES[i], float(v)) for v, i in zip(vals.tolist(), idxs.tolist())]
    return topk[0][0], topk[0][1], topk

# ------------------ Flask ------------------
app = Flask(__name__)

HTML = """
<!doctype html><html><head><meta charset="utf-8"><title>Waste Type</title>
<style>body{font-family:system-ui,Arial,sans-serif;max-width:760px;margin:40px auto;padding:0 16px;color:#0f172a}
.card{border:1px solid #e5e7eb;border-radius:12px;padding:16px;box-shadow:0 2px 10px rgba(0,0,0,.04)}
.row{display:flex;gap:16px;align-items:flex-start;flex-wrap:wrap}
img{max-width:320px;border-radius:8px;border:1px solid #e5e7eb}
table{border-collapse:collapse;width:100%;margin-top:12px}
th,td{padding:8px;border-bottom:1px solid #e5e7eb;text-align:left}
.btn{background:#0ea5e9;color:white;border:none;padding:10px 14px;border-radius:8px;cursor:pointer}
.btn:disabled{opacity:.6;cursor:not-allowed}
.pill{display:inline-block;padding:6px 12px;border-radius:999px;background:#e2f2ff;color:#0b67b8;font-weight:600}
.hint{color:#475569}</style></head><body>
<h2>Waste Type Classifier</h2>
<p class="hint">Labels: cardboard, glass, metal, paper, plastic, trash</p>
<div class="card">
  <form id="f" onsubmit="return false;">
    <input type="file" id="file" accept="image/*">
    <button class="btn" id="btn" onclick="predict()">Predict</button>
    <span id="msg" class="hint"></span>
  </form>
  <div id="pred" style="display:none;margin-top:12px;font-size:1.1rem">
    <strong>Predicted type:</strong> <span class="pill" id="top_lbl"></span>
    <span class="hint">(<span id="top_conf"></span> confidence)</span>
  </div>
  <div class="row" style="margin-top:12px">
    <img id="preview" src="" alt="">
    <div style="flex:1;min-width:260px">
      <table id="tbl" style="display:none"><thead><tr><th>Class</th><th>Probability</th></tr></thead><tbody></tbody></table>
    </div>
  </div>
</div>
<script>
const file=document.getElementById('file'),btn=document.getElementById('btn'),msg=document.getElementById('msg'),
preview=document.getElementById('preview'),predBox=document.getElementById('pred'),topLbl=document.getElementById('top_lbl'),
topConf=document.getElementById('top_conf'),tbl=document.getElementById('tbl'),tbody=tbl.querySelector('tbody');
file.addEventListener('change',()=>{const f=file.files[0]; if(!f) return; const r=new FileReader(); r.onload=e=>preview.src=e.target.result; r.readAsDataURL(f);});
function percent(x){return (x*100).toFixed(2)+'%';}
async function predict(){
  const f=file.files[0]; if(!f){msg.textContent="Choose an image first."; return;}
  btn.disabled=true; msg.textContent="Predicting...";
  const fd=new FormData(); fd.append('file', f);
  try{
    const res=await fetch('/predict_type',{method:'POST',body:fd}), ct=res.headers.get('content-type')||'', txt=await res.text();
    if(!ct.includes('application/json')) throw new Error('Non-JSON: '+txt.slice(0,120));
    const js=JSON.parse(txt); if(js.error){msg.textContent=js.error; btn.disabled=false; return;}
    topLbl.textContent=js.label; topConf.textContent=percent(js.confidence); predBox.style.display='';
    tbody.innerHTML=''; (js.topk||[]).forEach(([l,p])=>{const tr=document.createElement('tr'); const a=document.createElement('td'); a.textContent=l; const b=document.createElement('td'); b.textContent=percent(p); tr.appendChild(a); tr.appendChild(b); tbody.appendChild(tr);});
    tbl.style.display=''; msg.textContent='';
  }catch(e){ msg.textContent='Error: '+e; } finally{ btn.disabled=false; }
}
</script></body></html>
"""

@app.route("/", methods=["GET"])
def index(): return render_template_string(HTML)

@app.route("/healthz", methods=["GET"])
def health(): return jsonify(status="ok")

@app.route("/predict_type", methods=["POST"])
def predict_type():
    if "file" not in request.files or request.files["file"].filename == "":
        return jsonify(error='no file: send multipart/form-data with field "file"'), 400
    try:
        pil = Image.open(request.files["file"].stream).convert("RGB")
    except Exception:
        return jsonify(error="could not read image"), 400
    label, conf, topk = predict_pil(pil)
    return jsonify(label=label, confidence=conf, topk=topk)

# ---------- server + tunnel ----------
def pick_free_port(start=START_PORT, tries=50):
    p=start
    for _ in range(tries):
        s=socket.socket(socket.AF_INET, socket.SOCK_STREAM)
        try: s.bind(("0.0.0.0", p)); s.close(); return p
        except OSError: p+=1
    raise RuntimeError("No free port found")

def ensure_cloudflared():
    name = ("cloudflared-darwin-arm64" if "arm" in platform.machine().lower() and platform.system()=="Darwin"
            else "cloudflared-darwin-amd64" if platform.system()=="Darwin"
            else "cloudflared-windows-amd64.exe" if platform.system()=="Windows"
            else "cloudflared-linux-amd64")
    bin_dir = Path.home()/".local"/"bin"; bin_dir.mkdir(parents=True, exist_ok=True)
    path = bin_dir/("cloudflared.exe" if name.endswith(".exe") else "cloudflared")
    if path.exists(): return path
    url=f"https://github.com/cloudflare/cloudflared/releases/latest/download/{name}"
    print("Downloading cloudflared:", url)
    urllib.request.urlretrieve(url, path.as_posix())
    os.chmod(path, os.stat(path).st_mode | stat.S_IEXEC)
    print("✔ cloudflared at", path)
    return path

def start_tunnel(port:int, timeout:int=TUNNEL_TIMEOUT):
    try:
        bin_path = ensure_cloudflared()
    except Exception as e:
        print("⚠️ cloudflared download failed:", e)
        return ""
    cmd=[bin_path.as_posix(),"tunnel","--url",f"http://127.0.0.1:{port}","--no-autoupdate"]
    proc = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True, bufsize=1)
    # keep a global ref so it's not GC'ed
    globals()["_cf_proc"]=proc
    public=""
    t0=time.time()
    while time.time()-t0<timeout:
        line=proc.stdout.readline()
        if not line: break
        m=re.search(r"https://[a-z0-9-]+\.trycloudflare\.com", line)
        if m: public=m.group(0); break
    if public:
        print("🌍 Public URL:", public)
        return public
    else:
        print("⚠️ No trycloudflare URL yet; using local URL only.")
        # print a few recent log lines for clues
        for _ in range(10):
            try:
                ln=proc.stdout.readline().strip()
                if ln: print("cloudflared:", ln)
            except: break
        return ""

def run_waitress(p:int):
    serve(app, host="0.0.0.0", port=p)

PORT = pick_free_port()
threading.Thread(target=run_waitress, args=(PORT,), daemon=True).start()
print(f"Local URL:  http://127.0.0.1:{PORT}")
public = start_tunnel(PORT)

print("\n👉 Use THIS URL now:")
print("   ", public or f"http://127.0.0.1:{PORT}")
print("⚠️ If you restart this cell, you will get a NEW trycloudflare URL. Don’t reuse old ones.")
```
